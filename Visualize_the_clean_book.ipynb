 # Rating Analysis

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\deore\\FullStack Data Science course\\Book Review & Rating Analysis\\clean_books2.csv")
```

```python
plt.figure(figsize = (8,5))
plt.hist(df["Rating"],bins = 10)
plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Ratings")
plt.show()
```

# 2. Top Authors by Number of Books

```python
top_authors = df["Authors"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_authors.plot(kind="bar")
plt.xlabel("Author")
plt.ylabel("Number of Books")
plt.title("Top 10 Authors by Number of Books")
plt.xticks(rotation=45)
plt.show()
```

# 3. Publication Year

```python
# shows how many books were published each year?

books_per_year = df["Publication_year"].value_counts().sort_index()

plt.figure(figsize=(10,5))
books_per_year.plot(kind="line")
plt.xlabel("Publication Year")
plt.ylabel("Number of Books")
plt.title("Books Published Over the Years")
plt.show()
```

# 4. Rating vs Number of Reviews

```python
# shows does books with more reviews have higher ratings?

plt.figure(figsize=(8,5))
plt.scatter(df["Number_of_reviews"], df["Rating"])
plt.xlabel("Number of Reviews")
plt.ylabel("Rating")
plt.title("Rating vs Number of Reviews")
plt.show()
```

# 5. Rating vs Ratings Count

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Ratings_Count"], df["Rating"])
plt.xlabel("Ratings Count")
plt.ylabel("Rating")
plt.title("Rating vs Ratings Count")
plt.show()
```

# 6. Number of Reviews Distribution

```python
# How many reviews do books generally receive?

plt.figure(figsize=(8,5))
plt.hist(df["Number_of_reviews"], bins=30)
plt.xlabel("Number of Reviews")
plt.ylabel("Number of Books")
plt.title("Distribution of Number of Reviews")
plt.show()
```

```python
plt.figure(figsize=(8,4))
plt.boxplot(df["Number_of_reviews"].dropna())
plt.ylabel("Number of Reviews")
plt.title("Distribution of Number of Reviews")
plt.show()
```

# 7. Number of Pages

```python
plt.figure(figsize=(8,5))
plt.hist(df["Num_Of_Pages"].dropna(), bins=30)
plt.xlabel("Number of Pages")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Pages")
plt.show()
```

# 8. Publisher Analysis

```python
# Which publishers have the most books?

top_publishers = df["Publisher"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_publishers.plot(kind="bar")
plt.xlabel("Publisher")
plt.ylabel("Number of Books")
plt.title("Top 10 Publishers by Number of Books")
plt.xticks(rotation=45)
plt.show()
```

# 9. Language Analysis

```python
language_counts = df["Language_Code"].value_counts().head(10)

plt.figure(figsize=(8,5))
language_counts.plot(kind="bar")
plt.xlabel("Language")
plt.ylabel("Number of Books")
plt.title("Books by Language")
plt.show()
```
